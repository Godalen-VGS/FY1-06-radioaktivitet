Radioaktive kjerner henfaller **tilfeldig**. Vi kan aldri vite når én bestemt kjerne henfaller, men med mange kjerner blir mønsteret svært forutsigbart: antallet halveres for hver halveringstid.

I denne notebooken simulerer du henfall med tilfeldige tall, og analyserer måledata fra et vanlig skoleforsøk med barium-137m.

## Henfallsloven

Med $N_0$ kjerner ved start og halveringstid $T_{1/2}$ er antallet igjen etter tiden $t$

$$N(t) = N_0 \left(\tfrac{1}{2}\right)^{t/T_{1/2}} = N_0\, e^{-\lambda t} \qquad \text{der } \lambda = \frac{\ln 2}{T_{1/2}}$$

**Aktiviteten** $A = \lambda N$ er antall henfall per sekund, målt i becquerel (Bq).

## Simulering med terninger

Tenk deg at hver kjerne er en terning. I hvert tidssteg kaster vi alle terningene, og de som viser 6 «henfaller» og tas bort. Sannsynligheten for henfall per steg er da $p = \frac{1}{6}$.

`random.random()` gir et tilfeldig tall mellom 0 og 1. Er det mindre enn $p$, henfaller kjernen.

```python
import random
import matplotlib.pyplot as plt

N = 1000
p = 1/6
antall = [N]
for steg in range(20):
    N = sum(1 for _ in range(N) if random.random() >= p)
    antall.append(N)

plt.plot(antall, "o-")
plt.xlabel("Steg")
plt.ylabel("Antall kjerner igjen")
plt.grid(True)
plt.show()
```
